```python
# Program menampilkan huruf berdasarkan indeks

kalimat = "PRAKTIKUM ALPRO1"

print("String seluruhnya:", kalimat)
print("Huruf pertama:", kalimat[0])
print("Huruf terakhir:", kalimat[-1])

print("Huruf ke-2:", kalimat[1])
print("Huruf ke-3:", kalimat[2])
print("Huruf ke-5:", kalimat[4])
print("Huruf ke-4:", kalimat[3])
```

```python
# Program menukar nilai menggunakan temporary (temp)

x = 20
y = 10

print("Nilai awal x =", x, "y =", y)

temp = x
x = y
y = temp

print("Nilai setelah ditukar:")
print("Nilai x =", x)
print("Nilai y =", y)
```

```python
# Program operator bitwise

a = 20
b = 9

print("Nilai dari operasi bitwise AND antara a dan b =", a & b)
print("Nilai dari operasi bitwise OR antara a dan b =", a | b)
print("Nilai dari operasi bitwise XOR antara a dan b =", a ^ b)
```

```python
# Program kalkulasi midpoint

print("Program Kalkulasi Midpoint")

x1 = 2
y1 = 2
x2 = 4
y2 = 4

midpoint_x = (x1 + x2) / 2
midpoint_y = (y1 + y2) / 2

print("Nilai midpoint dari garis:")
print("Nilai midpoint x :", midpoint_x)
print("Nilai midpoint y :", midpoint_y)
```
