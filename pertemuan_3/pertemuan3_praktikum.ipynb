```python
#Kondisi if adalah kondisi yang akan dieksekusi oleh program jika bernilai benar atau trus
nilai = 9
#jika kondisi benar/true maka program akan mengeksekusi perintah dibawahnya
if(nilai > 7):
    print("sembilan lebih besar dari angka tujuh") #kondisi benar, di eksekusi
#jika kondisi salah/false maka program tidak akan mengeksekusi perintah dibawahnya 
if(nilai > 10):
    print("sembilan lebih besar dari angka sepuluh") #kondisi salah, maka tidak tereksekusi
```

```python
#Kondisi if else
nilai = 3

# Jika kondisi if bernilai True, kode if dijalankan.
# Jika False, kode else dijalankan.
if nilai > 7:
    print("selamat anda lulus")
else:
    print("maaf anda tidak lulus")
```

```python

# Data Mahasiswa G - Lana

nama = "Lana"
biaya_per_bulan = 800000

januari = "BLOKIR"
februari = "BUKA BLOKIR"
maret = "BLOKIR"
april = "-"
mei = "BUKA BLOKIR"
juni = "BLOKIR"
juli = "BUKA BLOKIR"

# Menghitung biaya buka blokir
jumlah_bulan = 0

if januari == "BUKA BLOKIR":
    jumlah_bulan += 1

if februari == "BUKA BLOKIR":
    jumlah_bulan += 1

if maret == "BUKA BLOKIR":
    jumlah_bulan += 1

if april == "BUKA BLOKIR":
    jumlah_bulan += 1

if mei == "BUKA BLOKIR":
    jumlah_bulan += 1

if juni == "BUKA BLOKIR":
    jumlah_bulan += 1

if juli == "BUKA BLOKIR":
    jumlah_bulan += 1

total = jumlah_bulan * biaya_per_bulan

print("Nama Mahasiswa:", nama)
print("Januari:", januari)
print("Februari:", februari)
print("Maret:", maret)
print("April:", april)
print("Mei:", mei)
print("Juni:", juni)
print("Juli:", juli)
print("Jumlah bulan buka blokir:", jumlah_bulan)
print("Total biaya: Rp", total)
```
